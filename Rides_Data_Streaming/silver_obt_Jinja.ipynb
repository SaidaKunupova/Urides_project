```python
from jinja2 import Template

jinja_str = """
select 

{% for config in jinja_config %}
    {{ config.select }} 
    {% if not loop.last %}
        ,
    {% endif %}
{% endfor %}
FROM
    {% for config in jinja_config %}
         {% if loop.first %}
            {{ config.table }}
        {% else %}
            LEFT JOIN {{ config.table }}
            ON {{ config.on }}
        {% endif %}
    {% endfor %}
        
    {% for config in jinja_config %}
        {% if loop.first%}
            {% if config.where != "" %}
            WHERE
            {% endif %}
        {% endif %}

        {{ config.where }}
            {% if not loop.last %}
                {% if config.where != "" %}
                AND
                {% endif %}
            {% endif %}
    {% endfor %}
    

 
 """
template = Template(jinja_str)
rendered_template = template.render(jinja_config=jinja_config)
# print(rendered_template)
dbutils.notebook.exit(rendered_template)
```
